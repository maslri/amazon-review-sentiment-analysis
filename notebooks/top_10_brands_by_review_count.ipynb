```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("../data/raw/train_data.csv")
products_df = pd.read_csv("../data/raw/title_brand.csv")
```

```python
reviews = df.merge(products_df, on="asin", how="left")
```

```python
reviews = reviews.dropna(subset=["brand"])
```

```python
brand_counts = reviews.groupby("brand").size().reset_index(name="review_count")
```

```python
top10_brands = brand_counts.sort_values(by="review_count", ascending=False).head(10)
```

```python
top_reviews = reviews[reviews["brand"].isin(top10_brands["brand"])]
```

```python
brand_rating = (
    top_reviews.groupby("brand")["overall"].mean().reset_index(name="average_rating")
)
```

```python
result = brand_rating.merge(top10_brands, on="brand")
```

```python
result = result.sort_values(by="average_rating", ascending=False)
result
```

```python
result = result[["brand", "average_rating"]].reset_index(drop=True)
result.index += 1
result
```
