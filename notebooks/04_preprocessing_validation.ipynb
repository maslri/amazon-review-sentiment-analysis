```python
import pandas as pd

from src.data_loader import load_train_reviews
from src.preprocessing.cleaner import (
    preprocess_for_tfidf,
    preprocess_for_transformer
)
```

```python
reviews = load_train_reviews()
```

```python
sample_reviews = reviews.sample(20, random_state=42).copy()
```

```python
sample_reviews["tfidf_text"] = sample_reviews["reviewText"].apply(preprocess_for_tfidf)

sample_reviews["transformer_text"] = sample_reviews["reviewText"].apply(
    preprocess_for_transformer
)
```
