```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("data/train_data.csv")
df.head()
```

```python
brand_df = pd.read_csv("data/title_brand.csv")
brand_df.head()
```

```python
five_star_reviews = df[df["overall"] == 5]
```

```python
top_products = (
    five_star_reviews.groupby("asin").size().reset_index(name="five_star_count")
)
```

```python
top_products = top_products.merge(brand_df, on="asin", how="left")
```

```python
top_products = top_products.sort_values(by="five_star_count", ascending=False)
```

```python
top10 = top_products.head(10)
```

```python
result = top10[["brand", "title", "five_star_count"]]

print(result)
```

```python
result.reset_index(drop=True, inplace=True)
result.index += 1
result
```
